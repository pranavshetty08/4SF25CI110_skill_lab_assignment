1. Counting Words in a File
Write a function that reads the contents of a file named document.txt and returns the number of words in the file.



```python
def count_words():
    file = open("document.txt", "r")

    content = file.read()

    words = content.split()

    total_words = len(words)

    print("Total number of words:", total_words)

    file.close()


count_words()
```

2. Finding and Replacing Text
Write a function that finds and replaces all occurrences of a given word in a file named data.txt with another word.

```python
def find_replace():
    file = open("data.txt", "r")

    content = file.read()

    file.close()

    old_word = input("Enter word to find: ")
    new_word = input("Enter replacement word: ")

    updated_content = content.replace(old_word, new_word)

    file = open("data.txt", "w")

    file.write(updated_content)

    file.close()

    print("Text replaced successfully!")

    print("\nUpdated File Content:")
    print(updated_content)


find_replace()
```

3. Reading a File in Reverse
Write a function that reads the contents of a file named reverse.txt and prints each line in reverse order.

```python
def reverse_lines():
    file = open("reverse.txt", "r")

    lines = file.readlines()

    file.close()

    print("File contents in reverse order:")

    for line in reversed(lines):
        print(line, end="")


reverse_lines()
```

4. Counting Lines, Words, and Characters
Write a function that reads the contents of a file named stats.txt and returns the number of lines, words, and characters in the file.



```python
def count_statistics():
    file = open("stats.txt", "r")

    content = file.read()

    file.close()

    lines = content.splitlines()
    words = content.split()
    characters = len(content)

    print("Total Lines:", len(lines))
    print("Total Words:", len(words))
    print("Total Characters:", characters)


count_statistics()
```
