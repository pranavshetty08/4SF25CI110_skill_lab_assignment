1. Array Creation and Manipulation
Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. Replace all the elements in the third column with 1.

```python
import numpy as np

# Create a 5x5 array with random integers
arr = np.random.randint(1, 21, size=(5, 5))

print("Original Array:")
print(arr)

# Replace all elements in the third column with 1
arr[:, 2] = 1

print("\nModified Array:")
print(arr)
```

2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.

```python
# Create an array from 1 to 16
arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

# Replace diagonal elements with 0
np.fill_diagonal(arr, 0)

print("\nModified Array:")
print(arr)
```

3. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.



```python
# Create a 5x5 array with random integers
arr = np.random.randint(1, 50, size=(5, 5))

print("Original Array:")
print(arr)

# Extract border elements
border = np.concatenate([
    arr[0, :],
    arr[-1, :],
    arr[1:-1, 0],
    arr[1:-1, -1]
])

print("\nBorder Elements:")
print(border)
```

4. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.



```python
# Create a 4x4 array
arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

# Calculate row-wise sum
row_sum = np.sum(arr, axis=1)

# Calculate column-wise sum
column_sum = np.sum(arr, axis=0)

print("\nRow-wise Sum:")
print(row_sum)

print("\nColumn-wise Sum:")
print(column_sum)
```

5. Create a NumPy array of shape (3, 3) with values from 1 to 9. Reshape the array to shape (1, 9) and then to shape (9, 1).



```python
# Create a 3x3 array
arr = np.arange(1, 10).reshape(3, 3)

print("Original Array:")
print(arr)

# Reshape to (1, 9)
arr1 = arr.reshape(1, 9)

print("\nArray of Shape (1, 9):")
print(arr1)

# Reshape to (9, 1)
arr2 = arr.reshape(9, 1)

print("\nArray of Shape (9, 1):")
print(arr2)
```

6. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age

```python
# Create a structured array
data_type = [
    ('name', 'U20'),
    ('age', 'i4'),
    ('weight', 'f4')
]

# Add data to the array
arr = np.array([
    ('Rahul', 22, 65.5),
    ('Anjali', 20, 52.3),
    ('Kiran', 25, 70.2),
    ('Priya', 21, 55.8),
    ('Arun', 19, 60.0)
], dtype=data_type)

print("Original Structured Array:")
print(arr)

# Sort the array by age
sorted_arr = np.sort(arr, order='age')

print("\nArray Sorted by Age:")
print(sorted_arr)
```
