1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

```python
import pandas as pd
import numpy as np

# Create a DataFrame with random integers
df = pd.DataFrame(
    np.random.randint(1, 20, size=(5, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Add a new column containing product of first two columns
df['Product'] = df['A'] * df['B']

print("\nDataFrame after adding Product column:")
print(df)
```

2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.



```python
# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 20, size=(4, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Calculate row-wise sum
row_sum = df.sum(axis=1)

# Calculate column-wise sum
column_sum = df.sum(axis=0)

print("\nRow-wise Sum:")
print(row_sum)

print("\nColumn-wise Sum:")
print(column_sum)
```

3. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 50, size=(5, 3)),
    columns=['A', 'B', 'C']
)

# Introduce NaN values
df.loc[1, 'A'] = np.nan
df.loc[2, 'B'] = np.nan
df.loc[3, 'C'] = np.nan

print("Original DataFrame:")
print(df)

# Fill NaN values with column mean
df = df.fillna(df.mean())

print("\nDataFrame after filling NaN values:")
print(df)
```

4. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

```python
# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 50, size=(6, 4)),
    columns=['A', 'B', 'C', 'D']
)

# Introduce NaN values
df.loc[1, 'A'] = np.nan
df.loc[3, 'C'] = np.nan
df.loc[4, 'B'] = np.nan

print("Original DataFrame:")
print(df)

# Drop rows containing NaN values
df = df.dropna()

print("\nDataFrame after dropping NaN rows:")
print(df)
```

5. Data Aggregation 

Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.



```python
# Create DataFrame
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})

print("Original DataFrame:")
print(df)

# Group by Category and calculate sum and mean
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("\nData Aggregation:")
print(result)
```

6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.



```python
# Create DataFrame
df = pd.DataFrame({
    'Product': np.random.choice(
        ['Laptop', 'Mobile', 'Tablet', 'Keyboard'],
        size=10
    ),
    'Category': np.random.choice(
        ['Electronics', 'Accessories'],
        size=10
    ),
    'Sales': np.random.randint(1000, 10000, size=10)
})

print("Original DataFrame:")
print(df)

# Group by Category and calculate total sales
result = df.groupby('Category')['Sales'].sum()

print("\nTotal Sales by Category:")
print(result)
```

7. Merging DataFrames

Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
# Create first DataFrame
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Rahul', 'Priya', 'Arun', 'Kiran']
})

# Create second DataFrame
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [85, 90, 78, 95]
})

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

# Merge DataFrames using common column
merged_df = pd.merge(df1, df2, on='ID')

print("\nMerged DataFrame:")
print(merged_df)
```

8. MultiIndex DataFrame

Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.



```python
# Create MultiIndex
index = pd.MultiIndex.from_tuples(
    [
        ('A', 1),
        ('A', 2),
        ('B', 1),
        ('B', 2),
        ('C', 1),
        ('C', 2)
    ],
    names=['Category', 'Number']
)

# Create DataFrame
df = pd.DataFrame(
    {'Value': [10, 20, 30, 40, 50, 60]},
    index=index
)

print("MultiIndex DataFrame:")
print(df)

# Access a particular category
print("\nCategory A:")
print(df.loc['A'])

# Access a particular element
print("\nElement at A, 1:")
print(df.loc[('A', 1)])

# Slice the DataFrame
print("\nSlicing A to B:")
print(df.loc[('A', 1):('B', 2)])
```

9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.



```python
# Create MultiIndex
index = pd.MultiIndex.from_product(
    [
        ['A', 'B', 'C'],
        ['X', 'Y']
    ],
    names=['Category', 'SubCategory']
)

# Create DataFrame with random values
df = pd.DataFrame(
    {'Value': np.random.randint(1, 100, size=6)},
    index=index
)

print("Original MultiIndex DataFrame:")
print(df)

# Calculate sum for each Category
print("\nSum by Category:")
print(df.groupby(level='Category').sum())

# Calculate sum for each SubCategory
print("\nSum by SubCategory:")
print(df.groupby(level='SubCategory').sum())

# Calculate sum for each Category and SubCategory
print("\nSum by Category and SubCategory:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

10. Applying Functions

Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 20, size=(5, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Create a function to double values
def double(x):
    return x * 2

# Apply function to DataFrame
result = df.map(double)

print("\nDataFrame after doubling values:")
print(result)
```

11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

```python
# Create DataFrame
df = pd.DataFrame(
    np.random.randint(1, 50, size=(6, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

# Apply lambda function to calculate row-wise sum
df['Total'] = df.apply(
    lambda row: row['A'] + row['B'] + row['C'],
    axis=1
)

print("\nDataFrame after adding Total column:")
print(df)
```

12. Working with Text Data

Create a Pandas Series with 5 random text strings. Convert all the strings to uppercase.

```python
# Create a Pandas Series
data = pd.Series([
    'python',
    'pandas',
    'numpy',
    'data science',
    'programming'
])

print("Original Series:")
print(data)

# Convert strings to uppercase
result = data.str.upper()

print("\nSeries after converting to uppercase:")
print(result)
```
